```python
# creating interaction tables
```

```python
%sql
create or replace view ext_novartis.raw.interactions as
select doctor_id,call_date as event_date from ext_novartis.raw.calls
union all
select doctor_id,email_date as event_date from ext_novartis.raw.emails
union all
select doctor_id,email_date as event_date from ext_novartis.raw.emails

```

```python
%sql
select * from interactions
```

### Doctor Engagement KPIs

#### Total Engagement per Doctor

```python
%sql
select doctor_id,count(distinct event_date) from interactions group by doctor_id order by count(distinct event_date)
```

#### Top 20 Influential Doctors

```python
%sql
select doctor_id,count(distinct event_date) from interactions group by doctor_id order by count(distinct event_date) limit 20
```

#### Engagement by Doctor Specialty

```python
%sql
select * from ext_novartis.raw.hcp
```

```python
%sql
select h.specialty,count(distinct h.doctor_id) from ext_novartis.raw.hcp h join ext_novartis.raw.interactions r on h.doctor_id=r.doctor_id group by h.specialty
```
